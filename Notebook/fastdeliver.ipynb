#### Fix missing delivery times

```python
# Convert delivery_time from string to time
# Keep the original data
orders_table['delivery_time_original'] = orders_table['delivery_time']

# Clean the delivery_time column
orders_table['delivery_time'] = pd.to_datetime(
    orders_table['delivery_time_original'],
    errors='coerce'
).dt.time

# Check missing values
print(
    f"There are {orders_table['delivery_time'].isna().sum()} null orders"
)
```

```python
# Find the most common delivery time
most_common_time = orders_table['delivery_time'].mode()[0]

# Fill missing values
orders_table['delivery_time'] = orders_table['delivery_time'].fillna(most_common_time)
```

#### Remove duplicate order IDs

```python
# Removing duplicate rows

print(f"Shape BEFORE cleaning: {orders_table.shape}")
print(f"Missing delivery_time:    {orders_table['delivery_time'].isnull().sum()}")
print(f"Duplicate rows:        {orders_table.duplicated().sum()}")
print(f"order_id dtype:       {orders_table['order_id'].dtype}")
print()

# ── Step 1: Remove duplicates ─────────────────────────────────────────────
orders_table_clean = orders_table.drop_duplicates()
print(f"After drop_duplicates: {orders_table_clean.shape[0]} rows (removed {orders_table.shape[0] - orders_table_clean.shape[0]})")
```

```python
orders_table_clean.info()
```

```python
# Convert pickup_time from string to time
orders_table['pickup_time'] = pd.to_datetime(
    orders_table['pickup_time'],
    errors='coerce'
).dt.time

# Check missing values
print(orders_table['pickup_time'].isna().sum())
```

```python
# most common pickup time
most_pickup_time = orders_table_clean['pickup_time'].mode()[0]

# Fill missing values
orders_table_clean['pickup_time'] = orders_table['pickup_time'].fillna(most_pickup_time)
```

```python
# convert order date to datetime format
orders_table_clean['order_date'] = pd.to_datetime(orders_table_clean['order_date'])
```

##### Convert order_value from strings to floats

```python
# Removing currency sign
 
orders_table_clean['order_value'] = (
    orders_table_clean['order_value']
    .astype(str)
    .str.replace('₦', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)

orders_table_clean['order_value'] = pd.to_numeric(orders_table_clean['order_value'], errors='coerce')
```

```python
orders_table_clean.describe()
```

```python
orders_table_clean['order_value'].fillna(orders_table_clean['order_value'].median().round(2))
```

#### Standardise status values

```python
# standardise status values
orders_table_clean['status'] = (
    orders_table_clean['status']
    .astype('string')
    .str.strip()
    .str.lower()
)
```
