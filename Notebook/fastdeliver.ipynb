#### Fix missing delivery times

```python
# Convert delivery_time from string to time
# Keep the original data
orders_table['delivery_time_original'] = orders_table['delivery_time']

# Clean the delivery_time column
orders_table['delivery_time'] = pd.to_datetime(
    orders_table['delivery_time_original'],
    errors='coerce'
).dt.time

# Check missing values
print(
    f"There are {orders_table['delivery_time'].isna().sum()} null orders"
)
```

```python
# Find the most common delivery time
most_common_time = orders_table['delivery_time'].mode()[0]

# Fill missing values
orders_table['delivery_time'] = orders_table['delivery_time'].fillna(most_common_time)
```

#### Remove duplicate order IDs

```python
# Removing duplicate rows

print(f"Shape BEFORE cleaning: {orders_table.shape}")
print(f"Missing delivery_time:    {orders_table['delivery_time'].isnull().sum()}")
print(f"Duplicate rows:        {orders_table.duplicated().sum()}")
print(f"order_id dtype:       {orders_table['order_id'].dtype}")
print()

# ── Step 1: Remove duplicates ─────────────────────────────────────────────
orders_table_clean = orders_table.drop_duplicates()
print(f"After drop_duplicates: {orders_table_clean.shape[0]} rows (removed {orders_table.shape[0] - orders_table_clean.shape[0]})")
```

##### Convert order_value from strings to floats

```python
# Removing currency sign
 
orders_table_clean['order_value'] = (
    orders_table_clean['order_value']
    .astype(str)
    .str.replace('₦', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)

orders_table_clean['order_value'] = pd.to_numeric(orders_table_clean['order_value'], errors='coerce')
```

```python
orders_table_clean['order_value'].fillna(orders_table_clean['order_value'].median().round(2))
```

#### Standardise status values

```python
# standardise status values
orders_table_clean['status'] = (
    orders_table_clean['status']
    .astype('string')
    .str.strip()
    .str.lower()
)
```

##### Build the analytical pipeline

```python
# Zone with highest revenue
zone_summary = (
    orders_table_clean
    .groupby("delivery_zone")
    .agg(
        total_revenue=("order_value", "sum"),
        order_count=("order_id", "count"),
        avg_order_value=("order_value", "mean")
    )
    .sort_values("total_revenue", ascending=False)
    .reset_index()
)
print(f"Zone summary: \n{zone_summary}")

revenue_by_zone = orders_table_clean.groupby("delivery_zone")["order_value"].sum()

top_zone = revenue_by_zone.idxmax()
top_revenue = revenue_by_zone.max()

print(f"The delivery zone that generates the most revenue is {top_zone}, with ₦{top_revenue:,.2f} in total revenue.")
```

```python
# Pick up zone with highest revenue
revenue_by_pickup_zone = orders_table_clean.groupby("pickup_zone")["order_value"].sum()
top_pickup_zone = revenue_by_pickup_zone.idxmax()
top_pickup_revenue = revenue_by_pickup_zone.max()

print(f"The pickup zone that generates the most revenue is {top_pickup_zone}, with ₦{top_pickup_revenue:,.2f} in total revenue.")
```

```python
# Total revenue by zone
orders_table_clean["total_revenue"] = orders_table_clean["order_value"] + orders_table_clean["delivery_fee"]

revenue_by_zone = (
    orders_table_clean
    .groupby("delivery_zone")["total_revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("TOTAL REVENUE BY ZONE (order value + delivery fee)")
print("-" * 50)
for zone, revenue in revenue_by_zone.items():
    print(f"{zone:<20} {revenue:>15,.2f}")

print()
top_zone = revenue_by_zone.idxmax()
top_revenue = revenue_by_zone.max()
print(f"→ Highest total revenue: {top_zone} ({top_revenue:,.2f})")
```

```python
# Highest failure rate by rider
failure_rate_by_rider = (
    orders_table_clean
    .groupby("rider_id")["status"]
    .apply(lambda x: (x == "failed").mean() * 100)
    .sort_values(ascending=False)
)

print(failure_rate_by_rider.head(10).round(1).astype(str) + "%")
print(f"→ The rider with the highest failure rate is {failure_rate_by_rider.idxmax()} with a failure rate of {failure_rate_by_rider.max():.1f}%.")
```

```python
pickup = pd.to_datetime(orders_table_clean["pickup_time"].astype(str), errors="coerce")
delivery = pd.to_datetime(orders_table_clean["delivery_time"].astype(str), errors="coerce")

orders_table_clean["duration_min"] = (delivery - pickup).dt.total_seconds() / 60

avg_delivery_time = orders_table_clean.loc[orders_table_clean["duration_min"] >= 0, "duration_min"].mean()
print(f"Average delivery time: {avg_delivery_time:.1f} minutes")
```

```python
# Summary of orders by delivery zone, rider, and payment method
clean_orders_summary = (
    orders_table_clean
    .groupby(["delivery_zone", "rider_id", "payment_method"])
    .size()
    .reset_index(name="order_count")
)
print(clean_orders_summary)
```

```python
avg_delivery_time_by_rider = (
    orders_table_clean
    [orders_table_clean["duration_min"] >= 0]
    .groupby("rider_id")["duration_min"]
    .mean()
    .sort_values(ascending=False)
    .reset_index(name="avg_delivery_time_min")
)

print(avg_delivery_time_by_rider.head(10))
```

```python
from sqlalchemy import create_engine, text
```

```python
## ----- Connect to PostgreSQL Database -----
engine = create_engine(
    'postgresql://postgres:password@localhost:5432/fastdeliver')

print("Connected to PostgreSQL database")

## ---- Write the cleaned DataFrame to PostgreSQL
orders_table_clean.to_sql('orders_table_clean', engine, if_exists='replace', index=False)

clean_orders_summary.to_sql('clean_orders_summary', engine, if_exists='replace', index=False)

print ("Tables written to database: ")
with engine.connect() as conn:
    tables = ["orders_table_clean", "clean_orders_summary"]
    for tbl in tables:
        
        count = conn.execute(text(f"SELECT COUNT(*) FROM {tbl}")).fetchone()[0]
        print(f" {tbl:<12} {count:>10} rows")
```
